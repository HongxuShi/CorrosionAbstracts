```python
import pandas as pd
import plotly.express as px

df=pd.read_csv("DATA.csv")
cols=["year","MWL","LC","LD","ASL","JD","HD"]#筛选
df_plot=df[cols]
df_plot["year"]=df_plot["year"].astype(str)
fig=px.parallel_coordinates(df_plot,
dimensions=["MWL","LC","LD","ASL","JD","HD"],
color="ASL",
labels={"MWL":"平均词长","LC":"词汇复杂度","LD":"词汇密度","ASL":"平均句长","JD":"术语密度","HD":"模糊词密度"},
title="腐蚀文献摘要语言学特征平行坐标图")

years=sorted(df_plot["year"].unique())
anu=[]

for y in years:
    anu.append(dict(label=y,method="update",args=[{"dimensions":[dict(label=col,values=df_plot[df_plot["year"]==y])
        for col in ["MWL","LC","LD","ASL","JD","HD"]]},{"title":f"{y}年腐蚀文献语言学特征平行坐标图"}]))

anu.insert(0,dict(label="近四年",method="update",args=[{"dimensions":[dict(label=col,values=df_plot.loc[:,col])
        for col in ["MWL","LC","LD","ASL","JD","HD"]]},{"title":"近四年腐蚀文献摘要语言学特征平行坐标图"}]))
fig.update_layout(updatemenus=[dict(type="dropdown",direction="down",x=0.55,y=1.4,buttons=anu)])
fig.show()
```
