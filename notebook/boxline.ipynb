```python
import matplotlib.pyplot as plt
import pandas as pd

data=pd.read_csv("DATA.csv")

LEILE1=data["MWL"]
LEILE2=data["LC"]
LEILE3=data["LD"]
LEILE4=data["ASL"]
LEILE5=data["JD"]
LEILE6=data["HD"]

plt.rcParams['font.sans-serif']=['SimHei']
plt.figure(figsize=(24,6))

#循环子图
data_list=[LEILE1, LEILE2, LEILE3, LEILE4, LEILE5, LEILE6]
name_list=["MWL","LC","LD","ASL","JD","HD"]

for i in range(6):
    plt.subplot(1,6,i+1)
    plt.boxplot(data_list[i],
                sym='+',showmeans=True,meanline=True,whis=1.5,
                boxprops={"color":"black"},
                flierprops={"marker":"x","markerfacecolor":"red","markersize":3},
                meanprops={"marker":"o","markerfacecolor":"red","markersize":3},
                medianprops={"linestyle":"-","color":"blue"})
    plt.title(name_list[i],fontsize=15,color="black")
    
```
