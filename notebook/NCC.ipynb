```python
import pandas as pd
df=pd.read_csv('dataset_with_abstract.csv')
mean_2023=df[df['year']==2023]['citations'].mean()
print(mean_2023)
mean_2024=df[df['year']==2024]['citations'].mean()
print(mean_2024)
mean_2025=df[df['year']==2025]['citations'].mean()
mean_2026=df[df['year']==2026]['citations'].mean()
print(mean_2025)
print(mean_2026)
df_mean=pd.DataFrame({'year':[2023,2024,2025,2026],'mean_citations':[mean_2023,mean_2024,mean_2025,mean_2026]})
print(df_mean)
df=pd.merge(df,df_mean,how='left',on='year')
df['NCC']=df['citations']/df['mean_citations']
print(df[['title', 'year', 'citations', 'NCC']].head())
df.to_csv('dataset_with_NCC.csv', index=False)
data=pd.read_csv('DATA.csv')
data['NCC']=df['NCC']
data.to_csv('DATA.csv', index=False)
```
