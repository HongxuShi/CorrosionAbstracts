```python
import pandas as pd
import plotly.express as px
df=pd.read_csv("DATA.csv")
df["year"]=df["year"].astype(str)
cols=["year","LD","ASL","JD","NCC","title"]
fig=px.scatter(
    df[cols],
    x="ASL",
    y="LD",
    size="NCC",#被引次数决定气泡大小
    color="JD",
    animation_frame="year",
    hover_name="title",#论文标题
    labels={"ASL":"平均句长",
            "LD":"词汇密度",
            "JD":"术语密度",
            "NCC":"标准化引用数"},
    title="腐蚀文献摘要语言学特征动态气泡演化图")
fig.update_traces(marker=dict(sizemin=5))
fig.show()
```
