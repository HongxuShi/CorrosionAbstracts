```python
import matplotlib.pyplot as plt
import pandas as pd

data=pd.read_csv("DATA.csv")
LEILE1=data["MWL"]
LEILE2=data["LC"]
LEILE3=data["LD"]
LEILE4=data["ASL"]
LEILE5=data["JD"]
LEILE6=data["HD"]

plt.rcParams['font.sans-serif']=['SimHei']
plt.figure(figsize=(18,12))

data_list=[LEILE1,LEILE2,LEILE3,LEILE4,LEILE5,LEILE6]
title_list=['平均词长','词汇复杂度','词汇密度','平均句长','术语密度','模糊词密度']

for i in range(6):
    plt.subplot(2,3,i+1)    
    plt.hist(data_list[i],
            bins=20,
            color="steelblue",
            alpha=0.5,
            edgecolor="black")
    plt.tick_params(axis='both',labelsize=12)#坐标轴刻度大小
    plt.title(title_list[i],fontsize=18,color="black")
plt.savefig("histograms.png")
plt.show()
```
